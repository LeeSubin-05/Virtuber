```python
import cv2
import numpy as np
import os
import json

# =========================================================
# 기본 경로 설정
# =========================================================

BASE_DIR = r"C:\Users\82106\OneDrive\바탕 화면\Vtube"
MODEL_JSON_PATH = os.path.join(BASE_DIR, "model.json")
PARTS_LOCATION_DIR = os.path.join(BASE_DIR, "parts_location")

# =========================================================
# 공통 설정
# =========================================================

CLEAR_OUTPUT_DIR = True
ALPHA_THRESHOLD = 0
MIN_AREA = 20

WHITE_THRESHOLD = 245
BLACK_THRESHOLD = 10

# =========================================================
# 모델 설정
# =========================================================

MODEL_CONFIGS = {
    "1": {
        "name": "parts",
        "input_path": os.path.join(BASE_DIR, "texture_00.png"),
        "json_path": os.path.join(PARTS_LOCATION_DIR, "parts_bbox.json"),
        "output_dir": os.path.join(BASE_DIR, "parts"),
        "json_canvas_width": 4096,
        "json_canvas_height": 4096,
        "background_mode": None,
        "use_connected_components": True,
    },
    "2": {
        "name": "parts_black",
        "input_path": os.path.join(BASE_DIR, "black.png"),
        "json_path": os.path.join(PARTS_LOCATION_DIR, "black_parts_bbox.json"),
        "output_dir": os.path.join(BASE_DIR, "parts_black"),
        "json_canvas_width": 4096,
        "json_canvas_height": 4096,
        "background_mode": "white",
        "use_connected_components": True,
    },
    "3": {
        "name": "parts_brown",
        "input_path": os.path.join(BASE_DIR, "brown.png"),
        "json_path": os.path.join(PARTS_LOCATION_DIR, "brown_parts_bbox.json"),
        "output_dir": os.path.join(BASE_DIR, "parts_brown"),
        "json_canvas_width": 8192,
        "json_canvas_height": 8192,
        "background_mode": "black",
        "use_connected_components": False,
    },
    "4": {
        "name": "parts_gray",
        "input_path": os.path.join(BASE_DIR, "gray.png"),
        "json_path": os.path.join(PARTS_LOCATION_DIR, "gray_parts_bbox.json"),
        "output_dir": os.path.join(BASE_DIR, "parts_gray"),
        "json_canvas_width": 8192,
        "json_canvas_height": 8192,
        "background_mode": "black",
        "use_connected_components": False,
    },
}

# =========================================================
# 수동 이름 변경표
# =========================================================

RENAME_MAPS = {
    

    "2": {
        "skin": {
        "part_000",
        "part_013",
        "part_002",
        "part_005",
        "part_009",
        "part_015",
        "part_018",
        "part_019",
        "part_031",
        "part_033",
        "part_012",
    },

    "hair": {
        "part_007",
        "part_011",
        "part_035",
        "part_036",
        "part_041",
        "part_043",
        "part_017",
        "part_027",
        "part_028",
        "part_032",
        
        "part_034",
        "part_037",
        "part_038",
        "part_039",
        "part_040",
    },

    "eye": {
        "part_008",
        "part_010",
        
        "part_029",
        "part_030",
        
        
        "part_025",
        "part_026",
    },

    "eye_back": {
        "part_014",
        "part_016",
        "part_022",
        "part_024",
        
    },

    "blusher": {
        "part_020",
        "part_021",
    },

    "mouth": {
        "part_001",
        
    },

    "cloth": {
        
        "part_003",
        "part_004",
        "part_006",
        "part_023",
        
        "part_042",
        "part_044",
        "part_045",
        "part_046",
    },

    "pin": {
    },

    "shadow": {
    },

    "3": {
        "skin": {
        "part_000",
        "part_003",
        "part_009",
        "part_010",
        "part_011",
        "part_012",
        "part_013",
        "part_014",
        "part_015",
        "part_016",
        "part_017",
        "part_018",
        "part_019",
        "part_020",
        "part_021",
        "part_022",
        "part_025",
        "part_026",
        "part_029",
        "part_030",
        "part_031",
        "part_032",
        "part_034",
        "part_035",
        "part_036",
        "part_037",
        "part_038",
        "part_039",
        "part_040",
        "part_043",
        "part_046",
        "part_057",
        "part_061",
        "part_064",
        "part_067",
        "part_072",
        "part_073",
        "part_075",
        "part_077",
        "part_080",
        "part_082",
        "part_083",
        "part_087",
        "part_090",
        "part_093",
        "part_099",
        "part_101",
        "part_105",
        "part_106",
        "part_109",
        "part_110",
        "part_113",
        "part_117",
        "part_119",
        "part_120",
        "part_121",
        "part_124",
        "part_131",
        "part_146",
        "part_155",
        "part_157",
        "part_159",
        "part_160",
        "part_162",
        "part_168",
        "part_169",
        "part_170",
        "part_192",
        "part_197",
    },

    "hair": {
        "part_001",
        "part_005",
        "part_006",
        "part_007",
        "part_008",
        "part_023",
        "part_033",
        "part_042",
        "part_047",
        "part_065",
        "part_068",
        "part_078",
        "part_084",
        "part_085",
        "part_089",
        "part_094",
        "part_100",
        "part_102",
        "part_103",
        "part_108",
        "part_122",
        "part_123",
        "part_125",
        "part_128",
        "part_129",
        "part_130",
        "part_135",
        "part_141",
        "part_143",
        "part_147",
        "part_150",
        "part_152",
        "part_154",
        "part_158",
        "part_196",
        "part_198",
    },

    "eye": {
        "part_050",
        "part_051",
        "part_052",
        "part_053",
        "part_054",
        "part_063",
        "part_066",
        "part_069",
        "part_071",
        "part_114",
        "part_174",
        "part_177",
        "part_191",
        "part_193",
        "part_194",
        "part_195",
        "part_204",
        "part_205",
        "part_207",
        "part_208",
    },

    "eye_back": {
        "part_139",
        "part_140",
        "part_142",
        "part_145",
        "part_172",
        "part_175",
        "part_206",
        "part_210",
    },

    "blusher": {
        "part_096",
        "part_098",
        "part_107",
        "part_136",
        "part_138",
        "part_144",
        "part_185",
        "part_186",
        "part_201",
        "part_203",
    },

    "mouth": {
        "part_048",
        "part_074",
        "part_088",
        "part_097",
        "part_133",
        "part_134",
        "part_151",
        "part_156",
        "part_164",
        "part_165",
        "part_179",
        "part_181",
        "part_182",
        "part_184",
        "part_188",
        "part_189",
        "part_190",
        "part_199",
        "part_200",
        "part_211",
    },

    "cloth": {
        "part_002",
        "part_004",
        "part_024",
        "part_027",
        "part_028",
        "part_041",
        "part_044",
        "part_045",
        "part_055",
        "part_056",
        "part_058",
        "part_079",
        "part_081",
        "part_086",
        "part_091",
        "part_111",
        "part_116",
        "part_118",
        "part_126",
        "part_127",
        "part_137",
        "part_148",
        "part_153",
        "part_161",
        "part_163",
        "part_166",
        "part_171",
        "part_173",
        "part_176",
        "part_178",
        "part_183",
        "part_202",
        "part_209",
        "part_212",
    },

    "pin": {
        "part_049",
        "part_059",
        "part_060",
        "part_062",
        "part_095",
        "part_104",
        "part_112",
        "part_115",
        "part_149",
    },

    "shadow": {
        "part_070",
        "part_076",
        "part_092",
        "part_132",
        "part_167",
        "part_180",
        "part_187",
    },
    },
    "4": {
        "skin": {
        "part_000",
        "part_006",
        "part_043",
        "part_049",
        "part_054",
        "part_057",
        "part_061",
        "part_069",
        "part_080",
        "part_100",
        "part_106",
        "part_108",
        "part_110",
        "part_117",
        "part_119",
        "part_120",
        "part_124",
        "part_125",
        "part_150",
        "part_153",
        "part_160",
        "part_162",
        "part_163",
        "part_183",
        "part_196",
        "part_216",
        "part_240",
        "part_241",
        "part_242",
        "part_243",
    },

    "hair": {
        "part_001",
        "part_003",
        "part_004",
        "part_005",
        "part_007",
        "part_011",
        "part_019",
        "part_024",
        "part_025",
        "part_027",
        "part_031",
        "part_035",
        "part_038",
        "part_046",
        "part_052",
        "part_053",
        "part_055",
        "part_056",
        "part_067",
        "part_076",
        "part_081",
        "part_092",
        "part_099",
        "part_105",
        "part_109",
        "part_158",
        "part_159",
        "part_161",
        "part_168",
        "part_172",
        "part_173",
        "part_174",
        "part_177",
        "part_239",
    },

    "eye": {
        "part_072",
        "part_073",
        "part_113",
        "part_130",
        "part_171",
        "part_182",
        "part_184",
        "part_185",
        "part_186",
        "part_188",
        "part_201",
        "part_211",
        "part_213",
        "part_217",
        "part_218",
        "part_219",
        "part_223",
        "part_224",
        "part_244",
        "part_245",
    },

    "eye_back": {
        "part_214",
        "part_215",
        "part_229",
        "part_233",
        "part_237",
        "part_238",
    },

    "blusher": {
        "part_009",
        "part_042",
        "part_122",
        "part_206",
        "part_207",
        "part_226",
        "part_227",
    },

    "mouth": {
        "part_012",
        "part_015",
        "part_016",
        "part_018",
        "part_022",
        "part_074",
        "part_123",
        "part_129",
        "part_136",
        "part_157",
        "part_164",
        "part_203",
        "part_228",
    },

    "cloth": {
        "part_002",
        "part_008",
        "part_013",
        "part_017",
        "part_020",
        "part_021",
        "part_023",
        "part_030",
        "part_040",
        "part_041",
        "part_044",
        "part_045",
        "part_047",
        "part_048",
        "part_050",
        "part_051",
        "part_058",
        "part_059",
        "part_063",
        "part_075",
        "part_077",
        "part_078",
        "part_082",
        "part_083",
        "part_084",
        "part_085",
        "part_086",
        "part_088",
        "part_089",
        "part_094",
        "part_097",
        "part_098",
        "part_101",
        "part_102",
        "part_111",
        "part_112",
        "part_114",
        "part_118",
        "part_127",
        "part_128",
        "part_133",
        "part_134",
        "part_138",
        "part_139",
        "part_140",
        "part_147",
        "part_148",
        "part_155",
        "part_156",
        "part_166",
        "part_169",
        "part_170",
        "part_176",
        "part_180",
        "part_181",
        "part_187",
        "part_189",
        "part_190",
        "part_191",
        "part_192",
        "part_194",
        "part_195",
        "part_198",
        "part_231",
        "part_232",
        "part_234",
        "part_235",
    },

    "pin": {
        "part_010",
        "part_014",
        "part_026",
        "part_028",
        "part_029",
        "part_032",
        "part_033",
        "part_034",
        "part_037",
        "part_039",
        "part_060",
        "part_062",
        "part_064",
        "part_065",
        "part_066",
        "part_068",
        "part_070",
        "part_071",
        "part_087",
        "part_090",
        "part_095",
        "part_096",
        "part_103",
        "part_107",
        "part_115",
        "part_116",
        "part_121",
        "part_126",
        "part_131",
        "part_132",
        "part_137",
        "part_141",
        "part_142",
        "part_143",
        "part_144",
        "part_145",
        "part_146",
        "part_149",
        "part_151",
        "part_152",
        "part_154",
        "part_165",
        "part_167",
        "part_175",
        "part_178",
        "part_179",
        "part_193",
        "part_197",
        "part_199",
        "part_200",
        "part_202",
        "part_204",
        "part_205",
        "part_208",
        "part_209",
        "part_210",
        "part_212",
        "part_220",
        "part_221",
        "part_222",
        "part_225",
        "part_230",
        "part_236",
    },

    "shadow": {
        "part_036",
        "part_079",
        "part_091",
        "part_104",
        "part_121",
        "part_135",
    },
    },
}
}

# =========================================================
# 1번 모델 추가 그룹
# =========================================================
# eye는 아래 번호만 eye로 저장
# 입 주변 파츠는 skin에 포함해서 얼굴과 같이 자연스럽게 색칠되도록 처리

MODEL1_EXTRA_GROUPS = {
    "skin": {
        "part_000",
        "part_001",
        "part_002",
        "part_003",
        "part_004",
        "part_018",
        "part_025",
        "part_026",
        "part_029",
        "part_030",
        "part_032",
        "part_033",
        "part_052",
        "part_072",
        "part_094",
        "part_102",
        "part_104",
        "part_106",
        "part_107",
        "part_108",
        "part_111",
        "part_112",
        "part_128",
        "part_129",
    },

    "eye": {
        "part_085",
        "part_086",
        "part_100",
        "part_101",
        "part_103",
        "part_105",
        "part_109",
        "part_110",
        "part_115",
        "part_116",
        "part_117",
        "part_119",
        "part_124",
        "part_125",
        "part_126",
    },

    "eye_back": {
        "part_083",
        "part_084",
        "part_087",
        "part_088",
        "part_091",
        "part_092",
        "part_120",
        "part_121",
        "part_130",
        "part_131",
    },

    "hair": {
        "part_055",
        "part_007",
        "part_027",
        "part_028",
        "part_037",
        "part_038",
        "part_048",
        "part_049",
        "part_053",
        "part_054",
        "part_056",
        "part_057",
        "part_058",
        "part_059",
        "part_060",
        "part_061",
        "part_062",
        "part_063",
        "part_064",
        "part_065",
        "part_066",
        "part_067",
        "part_068",
        "part_069",
        "part_070",
        "part_071",
    },

    "blusher": {
        
        "part_082",
        "part_127",
    },

    "cloth": {
        "part_005",
        "part_006",
        "part_010",
        "part_011",
        "part_012",
        "part_016",
        "part_017",
        "part_023",
        "part_024",
        "part_034",
        "part_035",
        "part_036",
        "part_039",
        "part_040",
        "part_041",
        "part_042",
        "part_043",
        "part_044",
        "part_073",
        "part_074",
    },

    "pin": {
        "part_008",
        "part_009",
        "part_013",
        "part_014",
        "part_019",
        "part_020",
        "part_021",
        "part_022",
        "part_045",
        "part_046",
        "part_047",
        "part_050",
        "part_051",
        "part_122",
        "part_123",
        "part_132",
        "part_133",
    },

    "mouth": {
        "part_015",
        "part_081",
        "part_095",
        "part_096",
        "part_097",
        "part_113",
        "part_114",
        "part_118",
        "part_134",
    },

    "eyebrow": {
        "part_075",
        "part_076",
        "part_077",
        "part_078",
        "part_079",
        "part_080",
        "part_089",
        "part_090",
        "part_098",
        "part_099",
    },

    "shadow": {
        "part_031",
        "part_093",
    },
}

# =========================================================
# 파일 읽기 / 쓰기
# =========================================================

def read_image_unchanged(path):
    if not os.path.exists(path):
        raise FileNotFoundError(f"이미지를 찾을 수 없습니다: {path}")

    data = np.fromfile(path, dtype=np.uint8)
    img = cv2.imdecode(data, cv2.IMREAD_UNCHANGED)

    if img is None:
        raise ValueError(f"이미지를 읽지 못했습니다: {path}")

    return img


def write_image(path, img):
    os.makedirs(os.path.dirname(path), exist_ok=True)

    ext = os.path.splitext(path)[1]
    if not ext:
        path += ".png"
        ext = ".png"

    success, encoded = cv2.imencode(ext, img)

    if not success:
        raise ValueError(f"이미지 인코딩 실패: {path}")

    encoded.tofile(path)


def load_parts_json(json_path):
    if not os.path.exists(json_path):
        raise FileNotFoundError(f"위치 JSON 파일이 없습니다: {json_path}")

    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    if isinstance(data, dict):
        if "extracted_parts" in data:
            data = data["extracted_parts"]
        elif "parts" in data:
            data = data["parts"]
        elif "items" in data:
            data = data["items"]
        elif "data" in data:
            data = data["data"]

    if not isinstance(data, list):
        raise ValueError(f"JSON 구조가 list 형식이 아닙니다: {json_path}")

    return data

# =========================================================
# 이미지 전처리
# =========================================================

def ensure_bgra(img, background_mode=None):
    if img is None:
        raise FileNotFoundError("이미지를 찾을 수 없습니다")

    if len(img.shape) == 3 and img.shape[2] == 4:
        return img

    if len(img.shape) == 3 and img.shape[2] == 3:
        b, g, r = cv2.split(img)

        if background_mode is None:
            raise ValueError("이 코드는 투명 PNG 기준입니다. 배경 제거가 필요하면 background_mode를 지정하세요")

        if background_mode == "white":
            bg_mask = (
                (r > WHITE_THRESHOLD)
                & (g > WHITE_THRESHOLD)
                & (b > WHITE_THRESHOLD)
            )
        elif background_mode == "black":
            bg_mask = (
                (r < BLACK_THRESHOLD)
                & (g < BLACK_THRESHOLD)
                & (b < BLACK_THRESHOLD)
            )
        else:
            raise ValueError(f"지원하지 않는 background_mode입니다: {background_mode}")

        alpha = np.where(bg_mask, 0, 255).astype(np.uint8)
        return cv2.merge([b, g, r, alpha])

    raise ValueError("지원하지 않는 이미지 형식입니다")

# =========================================================
# bbox / component 함수
# =========================================================

def bbox_iou(box_a, box_b):
    ax1, ay1, ax2, ay2 = box_a
    bx1, by1, bx2, by2 = box_b

    ix1 = max(ax1, bx1)
    iy1 = max(ay1, by1)
    ix2 = min(ax2, bx2)
    iy2 = min(ay2, by2)

    iw = max(0, ix2 - ix1)
    ih = max(0, iy2 - iy1)
    inter = iw * ih

    area_a = max(0, ax2 - ax1) * max(0, ay2 - ay1)
    area_b = max(0, bx2 - bx1) * max(0, by2 - by1)

    union = area_a + area_b - inter

    if union == 0:
        return 0

    return inter / union


def scale_bbox(bbox, sx, sy):
    x1, y1, x2, y2 = bbox
    return [
        int(round(x1 * sx)),
        int(round(y1 * sy)),
        int(round(x2 * sx)),
        int(round(y2 * sy)),
    ]


def clamp_bbox(bbox, w, h):
    x1, y1, x2, y2 = bbox

    x1 = max(0, min(w, x1))
    y1 = max(0, min(h, y1))
    x2 = max(0, min(w, x2))
    y2 = max(0, min(h, y2))

    return [x1, y1, x2, y2]


def get_scale(parts, w, h, json_canvas_width, json_canvas_height):
    json_max_x = max(part["bbox"][2] for part in parts)
    json_max_y = max(part["bbox"][3] for part in parts)

    if json_max_x > w or json_max_y > h:
        sx = w / json_canvas_width
        sy = h / json_canvas_height
    else:
        sx = 1.0
        sy = 1.0

    return sx, sy


def get_components(img):
    alpha = img[:, :, 3]
    mask = (alpha > ALPHA_THRESHOLD).astype(np.uint8) * 255

    num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(
        mask,
        connectivity=8,
    )

    components = []

    for label in range(1, num_labels):
        x, y, bw, bh, area = stats[label]

        if area < MIN_AREA:
            continue

        components.append({
            "label": label,
            "bbox": [int(x), int(y), int(x + bw), int(y + bh)],
            "width": int(bw),
            "height": int(bh),
            "area": int(area),
        })

    return labels, components

# =========================================================
# 폴더 / 이름 처리
# =========================================================

def clear_output_dir(output_dir):
    os.makedirs(output_dir, exist_ok=True)

    if not CLEAR_OUTPUT_DIR:
        return

    for filename in os.listdir(output_dir):
        if filename.lower().endswith((".png", ".json")):
            os.remove(os.path.join(output_dir, filename))


def make_unique_filename(output_dir, desired_name, part_id, used_names):
    base, ext = os.path.splitext(desired_name)

    if not ext:
        ext = ".png"

    candidate = base + ext

    if candidate not in used_names and not os.path.exists(os.path.join(output_dir, candidate)):
        return candidate

    part_number = str(part_id).replace("part_", "")
    candidate = f"{base}_{part_number}{ext}"

    count = 2
    while candidate in used_names or os.path.exists(os.path.join(output_dir, candidate)):
        candidate = f"{base}_{part_number}_{count}{ext}"
        count += 1

    return candidate


def read_model_choice_from_json(json_path):
    if not os.path.exists(json_path):
        print("model.json 파일이 없어서 직접 입력으로 진행합니다")
        choice = input("모델 번호 입력 1 parts 2 black 3 brown 4 gray: ").strip()
        return choice

    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    if isinstance(data, int):
        choice = str(data)
    elif isinstance(data, str):
        choice = data.strip()
    elif isinstance(data, dict):
        value = (
            data.get("model")
            or data.get("model_number")
            or data.get("choice")
            or data.get("result")
            or data.get("number")
        )
        if value is None:
            raise ValueError("model.json 안에서 model 번호를 찾지 못했습니다")
        choice = str(value).strip()
    else:
        raise ValueError("model.json 구조가 올바르지 않습니다")

    if choice not in MODEL_CONFIGS:
        raise ValueError(f"지원하지 않는 모델 번호입니다: {choice}")

    return choice

# =========================================================
# 자동 skin 판별 함수
# =========================================================

def get_visible_rgb_stats(crop):
    alpha = crop[:, :, 3]
    mask = alpha > ALPHA_THRESHOLD

    if not np.any(mask):
        return None

    rgb = crop[:, :, :3][:, :, ::-1]
    pixels = rgb[mask]

    mean_rgb = pixels.mean(axis=0)

    mean_rgb_uint8 = np.uint8([[mean_rgb]])
    hsv = cv2.cvtColor(mean_rgb_uint8, cv2.COLOR_RGB2HSV)[0, 0]
    h, s, v = int(hsv[0]), int(hsv[1]), int(hsv[2])

    return {
        "mean_r": float(mean_rgb[0]),
        "mean_g": float(mean_rgb[1]),
        "mean_b": float(mean_rgb[2]),
        "h": h,
        "s": s,
        "v": v,
    }


def looks_like_skin(crop):
    stats = get_visible_rgb_stats(crop)
    if stats is None:
        return False

    r = stats["mean_r"]
    g = stats["mean_g"]
    b = stats["mean_b"]
    h = stats["h"]
    s = stats["s"]
    v = stats["v"]

    cond1 = r > 180 and g > 130 and b > 110 and r >= g >= b and s < 95 and v > 150
    cond2 = (0 <= h <= 20 or 170 <= h <= 179) and 20 <= s <= 100 and v > 150 and r > g and g > b
    cond3 = r > 200 and g > 150 and b > 130 and (r - b) < 90

    return cond1 or cond2 or cond3



def find_manual_category(choice, part_id):
    """
    RENAME_MAPS / MODEL1_EXTRA_GROUPS 안에서 part_id가 어느 그룹에 속하는지 찾습니다.

    기존 코드의 오류 원인:
    - RENAME_MAPS는 {"skin": {"part_001", ...}} 구조인데
      기존 auto_name_part는 RENAME_MAPS[choice][part_id]처럼 찾고 있었습니다.
    - 그래서 수동 분류표를 넣어도 실제 파일명이 skin_001.png처럼 바뀌지 않았습니다.
    - 또한 기존 코드에서는 3번, 4번 모델이 2번 모델 안에 중첩되어 있어도
      아래 함수가 안전하게 찾아가도록 처리했습니다.
    """

    if choice == "1":
        source_map = MODEL1_EXTRA_GROUPS
    else:
        source_map = RENAME_MAPS.get(choice)

        # 혹시 3, 4 모델이 실수로 2번 모델 안에 들어간 구조도 복구해서 사용
        if source_map is None:
            source_map = RENAME_MAPS.get("2", {}).get(choice)

    if not isinstance(source_map, dict):
        return None

    for group_name, part_ids in source_map.items():
        # 잘못 중첩된 "3", "4" 같은 모델 번호는 그룹명이 아니므로 건너뜀
        if str(group_name) in MODEL_CONFIGS:
            continue

        if isinstance(part_ids, (set, list, tuple)) and part_id in part_ids:
            return str(group_name)

    return None


def auto_name_part(choice, part_id, old_part, crop):
    part_number = str(part_id).replace("part_", "")

    manual_category = find_manual_category(choice, part_id)
    if manual_category:
        return f"{manual_category}_{part_number}.png"

    original_name = old_part.get("file") or f"{part_id}.png"

    # 수동 분류표에 빠진 1번 모델 파츠만 보조적으로 skin 자동 판별
    if choice == "1" and looks_like_skin(crop):
        return f"skin_{part_number}.png"

    return original_name

# =========================================================
# 파츠 분리 실행
# =========================================================

def run_model(choice):
    config = MODEL_CONFIGS[choice]

    input_path = config["input_path"]
    json_path = config["json_path"]
    output_dir = config["output_dir"]

    print()
    print(f"[MODEL] {config['name']} 실행")
    print("원본 이미지:", input_path)
    print("위치 JSON:", json_path)
    print("저장 폴더:", output_dir)

    clear_output_dir(output_dir)

    removed_output_path = os.path.join(output_dir, "others.png")
    corrected_json_path = os.path.join(output_dir, "parts_bbox_corrected.json")

    img = read_image_unchanged(input_path)
    img = ensure_bgra(img, background_mode=config.get("background_mode"))

    h, w = img.shape[:2]
    print("원본 이미지 크기:", w, h)

    old_parts = load_parts_json(json_path)

    sx, sy = get_scale(
        old_parts,
        w,
        h,
        config["json_canvas_width"],
        config["json_canvas_height"],
    )

    print("좌표 스케일:", sx, sy)
    print("JSON 파츠 수:", len(old_parts))

    remaining_img = img.copy()
    corrected_parts = []
    used_names = set()

    labels = None
    components = []

    if config.get("use_connected_components", True):
        labels, components = get_components(img)
        print("찾은 연결 파츠 수:", len(components))

    for old_part in old_parts:
        part_id = old_part.get("id")
        if not part_id:
            continue

        expected_bbox = scale_bbox(old_part["bbox"], sx, sy)
        expected_bbox = clamp_bbox(expected_bbox, w, h)

        if config.get("use_connected_components", True):
            best_component = None
            best_iou = 0

            for comp in components:
                iou = bbox_iou(expected_bbox, comp["bbox"])
                if iou > best_iou:
                    best_iou = iou
                    best_component = comp

            if best_component is None:
                print("매칭 실패:", part_id)
                continue

            label = best_component["label"]
            x1, y1, x2, y2 = best_component["bbox"]

            crop = img[y1:y2, x1:x2].copy()
            component_mask = labels[y1:y2, x1:x2] == label

            crop_alpha = crop[:, :, 3].copy()
            crop[:, :, 3] = np.where(component_mask, crop_alpha, 0)

            real_area = int(np.count_nonzero(crop[:, :, 3] > ALPHA_THRESHOLD))

            remain_region = remaining_img[y1:y2, x1:x2]
            remain_region[component_mask] = [0, 0, 0, 0]
            remaining_img[y1:y2, x1:x2] = remain_region

            match_iou = round(float(best_iou), 4)

        else:
            x1, y1, x2, y2 = expected_bbox

            if x2 <= x1 or y2 <= y1:
                print("bbox 오류로 건너뜀:", part_id, expected_bbox)
                continue

            crop = img[y1:y2, x1:x2].copy()
            alpha = crop[:, :, 3]
            real_area = int(np.count_nonzero(alpha > ALPHA_THRESHOLD))

            if real_area < MIN_AREA:
                print("너무 작은 파츠라 건너뜀:", part_id, "area:", real_area)
                continue

            remaining_img[y1:y2, x1:x2, 3] = 0
            match_iou = None

        desired_name = auto_name_part(choice, part_id, old_part, crop)

        new_file_name = make_unique_filename(
            output_dir,
            desired_name,
            part_id,
            used_names,
        )

        save_path = os.path.join(output_dir, new_file_name)
        write_image(save_path, crop)
        used_names.add(new_file_name)

        corrected_info = {
            "id": part_id,
            "old_file": old_part.get("file", ""),
            "new_file": new_file_name,
            "bbox": [int(x1), int(y1), int(x2), int(y2)],
            "width": int(x2 - x1),
            "height": int(y2 - y1),
            "area": int(real_area),
            "bbox_area": int((x2 - x1) * (y2 - y1)),
            "match_iou": match_iou,
            "old_json_bbox": old_part["bbox"],
            "scaled_expected_bbox": expected_bbox,
        }

        corrected_parts.append(corrected_info)

        print(
            f"{part_id}.png -> {new_file_name} | "
            f"bbox={corrected_info['bbox']} | "
            f"width={corrected_info['width']} | "
            f"height={corrected_info['height']} | "
            f"area={corrected_info['area']} | "
            f"iou={corrected_info['match_iou']}"
        )

    with open(corrected_json_path, "w", encoding="utf-8") as f:
        json.dump(corrected_parts, f, ensure_ascii=False, indent=2)

    write_image(removed_output_path, remaining_img)

    print()
    print("완료")
    print("분리된 파츠 저장 폴더:", output_dir)
    print("수정된 bbox json:", corrected_json_path)
    print("나머지 이미지:", removed_output_path)

# =========================================================
# 실행
# =========================================================

def main():
    print("파츠를 하나로 합치지 않고 전부 따로 분리합니다")
    print("model.json 경로:", MODEL_JSON_PATH)

    choice = read_model_choice_from_json(MODEL_JSON_PATH)

    print("선택된 모델 번호:", choice)
    run_model(choice)


if __name__ == "__main__":
    main()
```
